```python
#Function Arguments
#Default arguments
#Required arguments
#Keyword arguments
#Variable-length arguments
```

```python
#Default arguments
def printinfo(name, age=25):
  print("Name:",name)
  print("Age",age)
```

```python
printinfo("Raj",21)
```

```python
printinfo("Raj")
```

```python
##Required arguments
def printinfo(name, age):
  print("Name:",name)
  print("Age",age)
```

```python
printinfo("raj")
```

```python
#Keyword arguments
def printinfo(name, age):
  print("Name:",name)
  print("Age",age)
```

```python
printinfo(age=21,name="Ravi")
```

```python
#4. Variable-length arguments
```

```python
def printinfo( arg1, *vartuple ):
  print ("Output is: ")
  print (arg1)
  for var in vartuple:
    print (var)
  return;
```

```python
printinfo( 10 )
printinfo( 70, 60, 50 ,100)
```

```python
#add number
def addany(a,*b):
  total=a
  for i in b:
    total=total+i
  return total
```

```python
addany(10,20,50)
```
