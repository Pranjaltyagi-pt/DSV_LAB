```python
import pandas as pd

data = {
    "Name": ["Pranjal", "Rahul", "Aman", "Neha"],
    "Age": [20, 21, 19, 20],
    "City": ["Agra", "Delhi", "Mathura", "Jaipur"]
}

student_df = pd.DataFrame(data)

print(student_df)
```

```python
import pandas as pd
import numpy as np

data = {
    "Book": ["Python Basics", "Data Science", "Java Programming", "AI Fundamentals"],
    "Author": ["John", "David", "James", "Robert"],
    "Price": [300, 450, 500, 600]
}

book_df = pd.DataFrame(data)

print(book_df)
```
