```python
#Question 4
import pandas as pd
import numpy as np

data = {
    "Name": ["Rahul", "Aman", "Priya", "Neha", "Rohit"],
    "Maths": [80, np.nan, 90, 70, np.nan],
    "Science": [75, 85, np.nan, 95, 80]
}

df = pd.DataFrame(data)
df.to_csv("student_marks.csv", index=False)

print("student_marks.csv created successfully!")
print(df)
```

```python
import pandas as pd

df = pd.read_csv("student_marks.csv")

print("Missing values before replacement:")
print(df.isnull().sum())

df.fillna(df.mean(numeric_only=True), inplace=True)

print("\nMissing values after replacement:")
print(df.isnull().sum())

print("\nUpdated DataFrame:")
print(df)
```

```python
#Question 8
import pandas as pd

attendance = pd.DataFrame({
    "RollNo": [101, 102, 103, 104],
    "Name": ["Rahul", "Aman", "Priya", "Neha"],
    "Attendance": [85, 90, 78, 95]
})

examination = pd.DataFrame({
    "RollNo": [101, 102, 103, 104],
    "Name": ["Rahul", "Aman", "Priya", "Neha"],
    "Marks": [80, 75, 92, 88]
})

with pd.ExcelWriter("college_data.xlsx") as writer:
    attendance.to_excel(writer, sheet_name="Attendance", index=False)
    examination.to_excel(writer, sheet_name="Examination", index=False)

print("Excel file created successfully!")
```

```python
import pandas as pd

excel_file = pd.ExcelFile("college_data.xlsx")

print("Available Sheet Names:")
print(excel_file.sheet_names)

df = pd.read_excel("college_data.xlsx", sheet_name="Attendance")

print("\nAttendance Records:")
print(df)
```

```python
#Question 9
import pandas as pd

students = pd.DataFrame({
    "RollNo": [101, 102, 103, 104],
    "Name": ["Rahul", "Aman", "Priya", "Neha"],
    "Course": ["B.Tech", "B.Tech", "B.Tech", "B.Tech"]
})

results = pd.DataFrame({
    "RollNo": [101, 102, 103, 104],
    "Subject": ["Python", "Python", "Python", "Python"],
    "Marks": [85, 76, 92, 88]
})

with pd.ExcelWriter("college_report.xlsx") as writer:
    students.to_excel(writer, sheet_name="Student_Details", index=False)
    results.to_excel(writer, sheet_name="Exam_Results", index=False)

print("college_report.xlsx created successfully!")
```

```python
#Question 10
import json

data = [
    {"RollNo": 101, "Name": "Rahul", "Marks": 85},
    {"RollNo": 102, "Name": "Aman", "Marks": 76},
    {"RollNo": 103, "Name": "Priya", "Marks": 92},
    {"RollNo": 104, "Name": "Neha", "Marks": 88}
]

with open("students.json", "w") as file:
    json.dump(data, file)

print("students.json created successfully!")
```

```python
import pandas as pd

df = pd.read_json("students.json")

print("Student Data:")
print(df)

print("\nData Types of Each Column:")
print(df.dtypes)
```
