```python
#web scrapping: https://quotes.toscrape.com/
import pandas as pd
```

```python
#import libraries
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://quotes.toscrape.com/"
```

```python
requests.get(url,timeout=15)
```

```python
response=requests.get(url,timeout=15)
```

```python
response.raise_for_status()
```

```python
s=BeautifulSoup(response.content,"html.parser")
```

```python
#extracting title
print("Title of website")
s.title.get_text(strip=True)
```

```python
#printing heading
print("Heading of website")
s.h1.get_text(strip=True)
```

```python

# Extract Quotes, Authors and Tags
data = []

for quote in s.find_all("div", class_="quote"):
    text = quote.find("span", class_="text").get_text(strip=True)

    author = quote.find("small", class_="author").get_text(strip=True)

    tags = [
        tag.get_text(strip=True)
        for tag in quote.find("div", class_="tags").find_all("a")
    ]

    data.append({
        "Quote": text,
        "Author": author,
        "Tags": tags
    })
```

```python
import pandas as pd

# Convert data into DataFrame
df = pd.DataFrame(data)

# Display DataFrame
print(df)

# Save DataFrame into CSV file
df.to_csv("quotes.csv", index=False)
```

```python
#Extract multiple page
data1=[]
for page in range(1,11):
  url=f"https://quotes.toscrape.com/page/{page}/"
  response=requests.get(url,timeout=15)
  soup=BeautifulSoup(response.content,"html.parser")
```

```python
import time
for quote in soup.select(".quote"):
    data1.append({
        "Page":page,
        "Quote":quote.select_one(".text").get_text(strip=True),
        "author":quote.select_one(".author").get_text(strip=True),

    })
    time.sleep(1)
```
